```python
from transformers import XLMRobertaTokenizer

# Load the tokenizer
tokenizer = XLMRobertaTokenizer.from_pretrained('xlm-roberta-base')

# Sample list of words
words = ["кристйанка","kristjanka", "obtoženkin", "обтоженкин", "njen", "ньен", "številen", "штевилен", "nasprotovanje", "наспротованье", "bogokletstvo", "богоклетство"]

# Tokenize each word and print tokens
for word in words:
    tokens = tokenizer.tokenize(word)
    print(f"Word: {word} --> Tokens: {tokens}")
```

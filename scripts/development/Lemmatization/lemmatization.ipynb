```python
#Lemmatization for Bulgarian with uppercase
import stanza

# Initialize the Bulgarian model for lemmatization
def ensure_bulgarian_model():
    try:
        print("Loading Bulgarian model...")
        nlp = stanza.Pipeline('bg', processors='tokenize,lemma')  # Tokenization and Lemmatization
        print("Bulgarian model loaded successfully.")
    except Exception as e:
        print(f"Error loading Bulgarian model: {e}")
        print("Attempting to download the Bulgarian model...")
        stanza.download('bg')  # Download Bulgarian model if it's not present
        nlp = stanza.Pipeline('bg', processors='tokenize,lemma')
    return nlp

def process_file(input_file, output_file):
    nlp = ensure_bulgarian_model()
    
    with open(input_file, 'r', encoding='utf-8') as infile, \
         open(output_file, 'w', encoding='utf-8') as outfile:
         
        lines = infile.readlines()
        sentence = []         # To collect sentence lines (here mainly the # text line)
        current_sent_id = None  # Track the last sent_id written
        
        for line in lines:
            stripped_line = line.strip()
            # New document: simply output and reset
            if stripped_line.startswith("# newdoc id"):
                # Process any pending sentence (if applicable)
                if sentence:
                    process_sentence(sentence, outfile)
                    sentence = []
                outfile.write(f"{stripped_line}\n")
                current_sent_id = None  # reset for new document
                
            # Sentence ID line
            elif stripped_line.startswith("# sent_id"):
                # Check if this sent_id is a duplicate of the one we already wrote.
                if current_sent_id is not None and stripped_line == current_sent_id:
                    # Skip duplicate sent_id
                    continue
                else:
                    if sentence:
                        # End previous sentence block by processing it.
                        process_sentence(sentence, outfile)
                        sentence = []
                    current_sent_id = stripped_line
                    outfile.write(f"{current_sent_id}\n")
                    
            # Text field line: keep the original text unchanged
            elif stripped_line.startswith("# text"):
                sentence.append(stripped_line)
                outfile.write(f"{stripped_line}\n")
            
            # Token lines (word and tag)
            elif stripped_line:
                token_data = stripped_line.split()
                if len(token_data) == 2:
                    token, tag = token_data
                    lemmatized_token = get_lemmatized_token_with_case(token, nlp)
                    outfile.write(f"{lemmatized_token}\t{tag}\n")
            
            # Blank line indicates end of a sentence block
            else:
                if sentence:
                    process_sentence(sentence, outfile)
                    sentence = []
        
        # Process any remaining sentence block at the end.
        if sentence:
            process_sentence(sentence, outfile)

def get_lemmatized_token_with_case(token, nlp):
    doc = nlp(token)
    original_word = token.strip()  # The original word
    lemmatized_word = doc.sentences[0].words[0].lemma
    
    # Preserve the case of the original word
    if original_word.istitle():
        return lemmatized_word.capitalize()  # Capitalize the lemmatized word
    elif original_word.isupper():
        return lemmatized_word.upper()  # Keep the lemmatized word in uppercase
    else:
        return lemmatized_word.lower()  # Keep it lowercase

def process_sentence(sentence, outfile):
    """
    This function is called at the end of a sentence block (when a blank line is encountered
    or before starting a new sentence). Instead of re-writing any collected metadata (like sent_id or # text)
    it simply writes two newlines.
    """
    outfile.write("\n\n")

# Input and output file paths
input_file_1 = "data/test/bg.txt"
input_file_2 = "data/train/bg.txt"
input_file_3 = "data/val/bg.txt"
output_file_1 = "data/test/lemmatized_bg.txt"
output_file_2 = "data/train/lemmatized_bg.txt"
output_file_3 = "data/val/lemmatized_bg.txt"
process_file(input_file_1, output_file_1)
process_file(input_file_2, output_file_2)
process_file(input_file_3, output_file_3)
```

```python
#working version without uppercase
import stanza

# Initialize the Bulgarian model for lemmatization
def ensure_bulgarian_model():
    try:
        print("Loading Bulgarian model...")
        nlp = stanza.Pipeline('bg', processors='tokenize,lemma')  # Tokenization and Lemmatization
        print("Bulgarian model loaded successfully.")
    except Exception as e:
        print(f"Error loading Bulgarian model: {e}")
        print("Attempting to download the Bulgarian model...")
        stanza.download('bg')  # Download Bulgarian model if it's not present
        nlp = stanza.Pipeline('bg', processors='tokenize,lemma')
    return nlp

def process_file(input_file, output_file):
    nlp = ensure_bulgarian_model()
    
    with open(input_file, 'r', encoding='utf-8') as infile, \
         open(output_file, 'w', encoding='utf-8') as outfile:
         
        lines = infile.readlines()
        sentence = []         # To collect sentence lines (here mainly the # text line)
        current_sent_id = None  # Track the last sent_id written
        
        for line in lines:
            stripped_line = line.strip()
            # New document: simply output and reset
            if stripped_line.startswith("# newdoc id"):
                # Process any pending sentence (if applicable)
                if sentence:
                    process_sentence(sentence, outfile)
                    sentence = []
                outfile.write(f"{stripped_line}\n")
                current_sent_id = None  # reset for new document
                
            # Sentence ID line
            elif stripped_line.startswith("# sent_id"):
                # Check if this sent_id is a duplicate of the one we already wrote.
                if current_sent_id is not None and stripped_line == current_sent_id:
                    # Skip duplicate sent_id
                    continue
                else:
                    if sentence:
                        # End previous sentence block by processing it.
                        process_sentence(sentence, outfile)
                        sentence = []
                    current_sent_id = stripped_line
                    outfile.write(f"{current_sent_id}\n")
                    
            # Text field line: process with lemmatizer and write updated text
            elif stripped_line.startswith("# text"):
                # Store the text line in our sentence collection (if needed)
                sentence.append(stripped_line)
                # Extract the actual text and lemmatize it.
                text = stripped_line.split("=", 1)[-1].strip()
                lemmatized_text = lemmatize_sentence(text, nlp)
                outfile.write(f"# text = {lemmatized_text}\n")
            
            # Token lines (word and tag)
            elif stripped_line:
                token_data = stripped_line.split()
                if len(token_data) == 2:
                    token, tag = token_data
                    lemmatized_token = get_lemmatized_token(token, nlp)
                    outfile.write(f"{lemmatized_token} {tag}\n")
            
            # Blank line indicates end of a sentence block
            else:
                if sentence:
                    process_sentence(sentence, outfile)
                    sentence = []
        
        # Process any remaining sentence block at the end.
        if sentence:
            process_sentence(sentence, outfile)

def lemmatize_sentence(text, nlp):
    doc = nlp(text)
    lemmatized_words = [word.lemma for sent in doc.sentences for word in sent.words]
    return " ".join(lemmatized_words)

def get_lemmatized_token(token, nlp):
    doc = nlp(token)
    return doc.sentences[0].words[0].lemma

def process_sentence(sentence, outfile):
    """
    This function is called at the end of a sentence block (when a blank line is encountered
    or before starting a new sentence). Instead of re-writing any collected metadata (like sent_id or # text)
    it simply writes two newlines.
    """
    outfile.write("\n\n")

# Input and output file paths
input_file = "bg.txt"
output_file = "updated_bg.txt"

process_file(input_file, output_file)
```

```python
import stanza

# Initialize Stanza pipelines for each language
bg_nlp = stanza.Pipeline('bg')  # Bulgarian
sl_nlp = stanza.Pipeline('sl')  # Slovenian
ru_nlp = stanza.Pipeline('ru')  # Russian
uk_nlp = stanza.Pipeline('uk')  # Ukrainian

# Define example sentences in each language
bg_text = "Той играеше футбол вчера."  # "He was playing football yesterday."
sl_text = "On je igral nogomet včeraj."  # "He was playing football yesterday."
ru_text = "Он играл в футбол вчера."  # "He was playing football yesterday."
uk_text = "Він грав у футбол вчора."  # "He was playing football yesterday."

# Process the sentences
bg_doc = bg_nlp(bg_text)
sl_doc = sl_nlp(sl_text)
ru_doc = ru_nlp(ru_text)
uk_doc = uk_nlp(uk_text)

# Function to extract lemmata
def extract_lemmata(doc):
    return [word.lemma for word in doc.words]

# Print lemmata for each sentence
print("Bulgarian Lemmas:", extract_lemmata(bg_doc))
print("Slovenian Lemmas:", extract_lemmata(sl_doc))
print("Russian Lemmas:", extract_lemmata(ru_doc))
print("Ukrainian Lemmas:", extract_lemmata(uk_doc))
```
